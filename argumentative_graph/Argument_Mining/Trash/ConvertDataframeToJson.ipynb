```python

#porzione di codice che definisce i vari argomenti presenti
arguments=[]
dict_tree={}
for i in range(len(df_all)):

    if df_all['prev_arg_id'][i]=="Depth0":

        dict_tree[i]=[df_all['arg_id'][i]]
        
    else:

        for key, value in dict_tree.items():
        
            if df_all['prev_arg_id'][i] in value:

                dict_tree[key].append(df_all['arg_id'][i])
```

```python
dict_arguments={}
dict_attacks={}

for i in dict_tree.keys():

    dict_arguments[i]=[]
    dict_attacks[i]=[]

#porzione di codice che definisce i vari argomenti presenti
arguments=[]

for i in range(len(df_all)):

    for key in dict_tree.keys():

        if df_all['arg_id'][i] in dict_tree[key]:

            dict_arguments[key].append({"id": df_all['arg_id'][i],"id_prev": df_all['prev_arg_id'][i], "sentences":[df_all['Argument'][i]], "acc": None})
```

```python
#porzione di codice che definisce gli attacchi presenti
attacks=[]

for i in range(len(df_all)):
    for j in range(len(df_all)):
        for key in dict_tree.keys():
    
            if df_all['arg_id'][i] in dict_tree[key] and df_all['prev_arg_id'][j] in dict_tree[key]:
    
                if (df_all['arg_id'][i] == df_all['prev_arg_id'][j]):
                    
                    dict_attacks[key].append({"source": df_all['arg_id'][j], "target": df_all['arg_id'][i], "weight": 1})
```
