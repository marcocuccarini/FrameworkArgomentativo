```python
import pandas as pd
depth1_file = '/Users/marco/Documents/GitHub/FrameworkArgomentativo/argumentative_graph/Argument_Graph/graph_depth1.csv'
depth1_df = pd.read_csv(depth1_file,encoding='latin-1')
depth1_df['depth'] = [1] * len(depth1_df)

depth2_file = '/Users/marco/Documents/GitHub/FrameworkArgomentativo/argumentative_graph/Argument_Graph/graph_depth2.csv'
depth2_df = pd.read_csv(depth2_file,encoding='latin-1')
depth2_df['depth'] = [2] * len(depth2_df)

depth3_file = '/Users/marco/Documents/GitHub/FrameworkArgomentativo/argumentative_graph/Argument_Graph/graph_depth3.csv'
depth3_df = pd.read_csv(depth3_file, encoding='latin-1')
depth3_df['depth'] = [3] * len(depth3_df)

depth4_file = '/Users/marco/Documents/GitHub/FrameworkArgomentativo/argumentative_graph/Argument_Graph/graph_depth4.csv'
depth4_df = pd.read_csv(depth4_file,encoding='latin-1')
depth4_df['depth'] = [4] * len(depth4_df)

depth5_file = '/Users/marco/Documents/GitHub/FrameworkArgomentativo/argumentative_graph/Argument_Graph/graph_depth5.csv'
depth5_df = pd.read_csv(depth5_file,encoding='latin-1')
depth5_df['depth'] = [5] * len(depth5_df)

frames = [depth1_df, depth2_df, depth3_df, depth4_df, depth5_df]

```

```python
df_all = pd.concat(frames)
arg_dic = dict(zip(df_all.arg_id, df_all.Argument))
arg_dic['a0'] = 'University fees in the UK should be kept at 9k'

df_all.reset_index(inplace=True)

df_all.to_csv("data.csv")
```

```python

#porzione di codice che definisce i vari argomenti presenti
arguments=[]
dict_tree={}
for i in range(len(df_all)):

    if df_all['prev_arg_id'][i]=="Depth0":

        dict_tree[i]=[df_all['arg_id'][i]]
        
    else:

        for key, value in dict_tree.items():
        
            if df_all['prev_arg_id'][i] in value:

                dict_tree[key].append(df_all['arg_id'][i])
```

```python
dict_arguments={}
dict_attacks={}

for i in dict_tree.keys():

    dict_arguments[i]=[]
    dict_attacks[i]=[]

#porzione di codice che definisce i vari argomenti presenti
arguments=[]

for i in range(len(df_all)):

    for key in dict_tree.keys():

        if df_all[i]['arg_id'] in dict_tree[key]:

            dict_arguments[key].append({"id": df_all['arg_id'][i],"id_prev": df_all['prev_arg_id'][i], "sentences":[df_all['Argument'][i]], "acc": "true"})
```

```python
#porzione di codice che definisce gli attacchi presenti
attacks=[]
for i in range(len(df_all)):
    for j in range(len(df_all)):

        if (df_all['arg_id'][i] == df_all['prev_arg_id'][j]):
            
            attacks.append({"source": df_all['arg_id'][j], "target": df_all['arg_id'][i], "weight": 1})
    
```

```python
import pandas as pd



# Step 1: Add a group column (e.g., group every 5 rows)
df_all['group'] = df_all.index // 5  # groups of 5

# Step 2: Group into nested dictionary
nested_dict = df_all.groupby('group').apply(
    lambda group: [
        {
            "id": row['arg_id'],
            "id_prev": row['prev_arg_id'],
            "sentences": [row['Argument']],
            "acc": "true"
        }
        for _, row in group.iterrows()
    ]
).to_dict()

# Step 3: Wrap in an outer dict (optional)
result = {
    str(key): value for key, value in nested_dict.items()
}

# Step 4: (Optional) Print result
import json
print(json.dumps(result, indent=4))
```
